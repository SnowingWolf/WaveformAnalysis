# DAQAnalyzer 缓存状态显示测试

测试新增的缓存状态显示功能

```python
from waveform_analysis.utils.daq.daq_analyzer import DAQAnalyzer
```

## 1. 扫描 DAQ 数据

```python
da = DAQAnalyzer(
    daq_root="/mnt/data/Run3/DAQ",
    daq_adapter="vx2730"
)
da.scan_all_runs()
print(f"找到 {len(da.runs)} 个运行")
```

## 2. 不显示缓存列（向后兼容）

```python
da.display_overview(sort_by="name", max_rows=10)
```

## 3. 显示缓存状态（默认数据类型）

```python
da.display_overview(
    sort_by="name",
    max_rows=10,
    show_cache=True,
    cache_storage_dir="./strax_data"
)
```

## 4. 自定义缓存数据类型

```python
da.display_overview(
    sort_by="name",
    max_rows=10,
    show_cache=True,
    cache_storage_dir="./strax_data",
    cache_data_types=["st_waveforms", "basic_features", "pulse_width"]
)
```

## 5. 按大小排序并显示缓存

```python
da.display_overview(
    sort_by="size",
    ascending=False,
    max_rows=15,
    show_cache=True,
    cache_storage_dir="./strax_data",
    cache_data_types=["st_waveforms", "basic_features"]
)
```
